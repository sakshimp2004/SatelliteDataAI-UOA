(2) The Landsat 7 data displays as red colored/tinged image. Why is this? [Think about both the content in Lecture 1, and what is going on when you visualize a satellite image]. (2 pts)

The Landsat 7 data displays red colored/tinged image because the Near-Infrared (NIR) band is mapped to the Red color channel of the display.

Key Reasons for the Red Appearance :-

1). Vegetation Reflectance : Healthy vegetation reflects electromagnetic radiation very strongly in the Near-Infrared (NIR) region of the spectrum, far more than it reflects visible green light.

2). False-Color Composite (RGB Mapping) :- To visualize this invisible infrared radiation on a computer screen, as standard false-color composite (typically Bands 4,3,2 for Landsat 7) is used.

3). Channel Assignment :- Band 4 (NIR) is assigned to the Red display channel, Band 3 (Red) to the Green channel, and Band 2 (Green) to the Blue channel.

4). Resulting Color :- Because vegetation reflects the highest amount of energy in the NIR band, the Red channel dominates the visual output, causing densely vegetated areas to appear bright red or red-tinged.

# Spectral reflectance curves
Next we we will generate a spectral reflectance curve for a given area.
This is a useful tool when you are considering if a given target has spectral characteristics for identification in ML. If everything has the same spectral signature, then you need to take a different approach to a simple spectral classification, as might have been the first instinct of a classical remote sensing scientist before the advent of machine learning.

```python
# Add in a library for plotting the data we extract
import matplotlib.pyplot as plt
```

```python
# Define a point and a small rectangular region around it
point = ee.Geometry.Point([174.7633, -36.8485])  # Auckland, NZ
region = point.buffer(150).bounds()  # ~300 m box
```

```python
# Load a Landsat 7 Collection 2 Level 2 SR image (surface reflectance)
image = ee.ImageCollection('LANDSAT/LE07/C02/T1_L2') \
    .filterBounds(region) \
    .filterDate('2002-01-01', '2002-12-31') \
    .sort('CLOUD_COVER') \
    .first()
```

(4) Why are we using the surface reflectance (SR) here rather than the top of atmosphere (ToA)? (2 pt)

1). Atmosphere Correction :- SR removes the scattering and absorption effects caused by gases, water vapour and aerosols in Earth's atmosphere.

2). True Surface Colors :- It measures light reflected directly by ground features instead of measuring what reaches the satellite sensor.

3). Consistency :- It allows you to compare multiple images over time or across different regions under changing weather and lightning conditions.

```python
# Select surface reflectance bands and apply scale factor
bands = ['SR_B1', 'SR_B2', 'SR_B3', 'SR_B4', 'SR_B5', 'SR_B7']
band_names = {
    'SR_B1': 'Blue (0.45–0.52 μm)',
    'SR_B2': 'Green (0.52–0.60 μm)',
    'SR_B3': 'Red (0.63–0.69 μm)',
    'SR_B4': 'NIR (0.77–0.90 μm)',
    'SR_B5': 'SWIR1 (1.55–1.75 μm)',
    'SR_B7': 'SWIR2 (2.09–2.35 μm)'
}

scale_factor = 2.75e-5  # From USGS documentation for Landsat C2 L2
```

(5) What is a 'scale factor' as has been used in the prior code cell, and why are they used in satellite data storage-to-processing workflows such as this? (2 pts)

A scale factor is a multiplier used to convert compressed, packed integer data back into true scientific decimal values.

1). Data Compression :- Satellite imagery stores data as integers (like 16-bit integers) to dramatically reduce file sizes and save storage space.

2). Processing Conversion :- Decimal values (floating-point numbers between 0 and 1 for reflectance) take up too much storage memory.

3). The Formula :- The software multiplies the stored integer by the scale factor (e.g. 2.75e-5) during processing to restore the actual physical measurements.

**Important sidenote**

The 'reduce' operation is a key concept in the backend of how Google is storing and serving us large quanities of satellite data. GEE uses 'reducers' in multiple ways, many of which are not always immediately intuitive. Check out the docs for more information:

*   https://developers.google.com/earth-engine/guides/reducers_intro

Here we are using a reduce operation over the specified small sample area in order to find the mean of all the cells that lie within the region:

```python
# Reduce region to mean reflectance
mean_dict = image.select(bands).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=region,
    scale=30,
    maxPixels=1e6
)
```

```python
# Extract and scale values
mean_values = mean_dict.getInfo()
reflectance = [mean_values[b] * scale_factor for b in bands]

# Plot the spectral curve
plt.figure(figsize=(10, 5))
plt.plot(list(band_names.values()), reflectance, marker='o', linestyle='-')
plt.title('Spectral Reflectance Curve (Landsat 7)')
plt.ylabel('Reflectance')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

(6) If we want to apply multiple reducers to the same inputs what is a more efficient way to do so than just writing out a long series of sequential reducers on our end? (2 pts)

In Google Earth Engine, the most efficient approach is to combine multiple reducers using .combine() on an ee.Reducer object.

Efficiency Benefit: Combining reducers allows Google Earth Engine's server-side computation engine to evaluate all required statistics in a single parallel pass over the underlying image pixels, drastically reducing network overhead, memory overhead, and execution latency compared to running multiple independent .reduceRegion() calls sequentially.

```python
# Combining mean, standard deviation, and min/max into a single execution pass
combined_reducer = (
    ee.Reducer.mean()
    .combine(reducer2=ee.Reducer.stdDev(), sharedInputs=True)
    .combine(reducer2=ee.Reducer.minMax(), sharedInputs=True)
)

# Execute a single reduceRegion call over the server backend
stats = image.reduceRegion(
    reducer=combined_reducer, geometry=region, scale=30
)
```

# Cloud cover
A final consideration that I want to cover in this lab is that of cloud cover. Cloud cover removal is a complex and often difficut process that GISCI 341 students spend an entire two weeks tackling as a focus (so use them in your teams!). However, for our purpose here lets just use the basic in-built cloud clearing native to the Landsat 7 collection on GEE.

```python
# Define small region around Auckland, NZ
point = ee.Geometry.Point([174.7633, -36.8485])
region = point.buffer(500).bounds()

# Load the image with least cloud cover in 2002
image = ee.ImageCollection("LANDSAT/LE07/C02/T1_L2") \
    .filterBounds(region) \
    .filterDate("2002-01-01", "2002-12-31") \
    .sort("CLOUD_COVER") \
    .first() #<- note the filtering step that is being applied here! We are taking the first image in a sorted by cloud cover response.

# Check metadata
print('Image ID:', image.get('LANDSAT_PRODUCT_ID').getInfo())
print('Cloud Cover (whole scene):', image.get('CLOUD_COVER').getInfo())
```

```python
# Function to mask clouds using QA_PIXEL band
def mask_clouds(image):
    qa = image.select('QA_PIXEL')
    # Bits 3 and 4 are cloud and cloud shadow respectively
    cloud_mask = qa.bitwiseAnd(1 << 3).eq(0).And(
                 qa.bitwiseAnd(1 << 4).eq(0))
    return image.updateMask(cloud_mask)

# Apply mask
masked_image = mask_clouds(image)
```

**Important side note**

In the function above we are using a bitwise operation in order to search through the quality assurance (QA) band of the image. Different satellites have different bit-masks. Read more about bitmasks and how they are used here:
*   https://spatialthoughts.com/2021/08/19/qa-bands-bitmasks-gee/



**Another important side note**

You will have noticed that we are frequently running code and not seeing anything happen in terms of output in the cell or changes in the RAM/Disk of the Colab instance (check the small graphs top right). This is deliberate! Satellite data is large and hard to wrangle. GEE does a lot of the work for us, hence we can explore and get into it so quickly. Part of how it does this is by not executing any of your API calls until you specifically ask for an output to be sent back to you. Prior to calling an output (usually through a 'AddLayer' or similar call), when using the GEE/EE API you are essentially chaining together a lot of dot-operators (e.g. '.sort \ .reduce') into a series of instructions that are verified, but only then acted on when required.

This is the result of a 'client vs. server' approach. For more detail, that will help you understand why you can and cannot do certain things, see:
*   https://developers.google.com/earth-engine/guides/client_server



```python
# Report Image Quality Over ROI
# Total pixels in region
pixel_area = ee.Image.pixelArea().clip(region)

# Count valid pixels before and after masking
total_pixels = pixel_area.reduceRegion(
    reducer=ee.Reducer.count(),
    geometry=region,
    scale=30,
    maxPixels=1e9
).getInfo()['area']

valid_pixels = masked_image.select('SR_B3').reduceRegion(
    reducer=ee.Reducer.count(),
    geometry=region,
    scale=30,
    maxPixels=1e9
).getInfo()['SR_B3']

cloud_coverage_pct = 100 * (1 - valid_pixels / total_pixels)

print(f"Total Pixels: {total_pixels}")
print(f"Valid (non-cloud) Pixels: {valid_pixels}")
print(f"Estimated Cloud Coverage in ROI: {cloud_coverage_pct:.2f}%")
```

```python
# Visualize the before and after
Map = geemap.Map(center=[-36.8485, 174.7633], zoom=12)

vis = {
    'bands': ['SR_B3', 'SR_B2', 'SR_B1'],
    'min': 0,
    'max': 30000,
    'gamma': 1.4
}

Map.addLayer(image, vis, 'Original Image (Cloudy)')
Map.addLayer(masked_image, vis, 'Cloud Masked')
Map
```

(7) Produce a table that tells me the average amount of cloud cover per LandSat image in a year: for 2005, 2010 and 2015. Do this over a region of your choice, as a percentage.

Provide a notebook example of your code that does so, in addition to your table. (5 pts)

```python
import ee

# ee.Initialize(project="earthengine-ml-testing")
ee.Initialize(project='earthenginetest-504004')

# Define region of interest (ROI): Auckland Region
roi = ee.Geometry.Rectangle([174.4, -37.1, 175.2, -36.5])

years = [2005, 2010, 2015]
cloud_results = {}

for year in years:
    # Query Landsat 7 Collection 2 Level 2 Tier 1 image collection over the target year
    collection = (
        ee.ImageCollection("LANDSAT/LE07/C02/T1_L2")
        .filterBounds(roi)
        .filterDate(f"{year}-01-01", f"{year}-12-31")
    )

    # Compute mean scene cloud cover from image metadata property
    avg_cloud = collection.aggregate_mean("CLOUD_COVER").getInfo()
    cloud_results[year] = avg_cloud

# Output Results
print("Year | Average Scene Cloud Cover (%)")
print("-----------------------------------")
for yr, cov in cloud_results.items():
    print(f"{yr} | {cov:.2f}%")
```

```python
# Additional required packages
import matplotlib.patches as patches
import urllib.request
import tempfile
import matplotlib.image as mpimg

# Define point and regions
point = ee.Geometry.Point([174.7633, -36.8485])
region = point.buffer(150).bounds()        # ~300 m box
expanded_region = point.buffer(1000).bounds()  # box for image preview

# Load Landsat 7 surface reflectance image
image = ee.ImageCollection('LANDSAT/LE07/C02/T1_L2') \
    .filterBounds(region) \
    .filterDate('2002-01-01', '2002-12-31') \
    .sort('CLOUD_COVER') \
    .first()

# Define bands and scaling
bands = ['SR_B1', 'SR_B2', 'SR_B3', 'SR_B4', 'SR_B5', 'SR_B7']
band_names = {
    'SR_B1': 'Blue (0.45–0.52 μm)',
    'SR_B2': 'Green (0.52–0.60 μm)',
    'SR_B3': 'Red (0.63–0.69 μm)',
    'SR_B4': 'NIR (0.77–0.90 μm)',
    'SR_B5': 'SWIR1 (1.55–1.75 μm)',
    'SR_B7': 'SWIR2 (2.09–2.35 μm)'
}
scale_factor = 2.75e-5

# Calculate mean reflectance
mean_dict = image.select(bands).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=region,
    scale=30,
    maxPixels=1e6).getInfo()

reflectance = [mean_dict[b] * scale_factor for b in bands]
labels = list(band_names.values())

# Get RGB thumbnail slightly zoomed out
thumb_params = {
    'dimensions': 512,
    'region': expanded_region,
    'format': 'png',
    'bands': ['SR_B3', 'SR_B2', 'SR_B1'],
    'min': 0,
    'max': 30000,
    'gamma': 1.4
}
url = image.getThumbURL(thumb_params)

# Get coordinates of regions
def get_bounds_coords(bounds):
    coords = bounds.coordinates().getInfo()[0]
    lons = [pt[0] for pt in coords]
    lats = [pt[1] for pt in coords]
    return min(lons), max(lons), min(lats), max(lats)

# Extract bounding boxes
xmin, xmax, ymin, ymax = get_bounds_coords(expanded_region)
rxmin, rxmax, rymin, rymax = get_bounds_coords(region)

# Plot both panels
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), gridspec_kw={'height_ratios': [2, 1]})

# Load and plot image with annotation
with tempfile.NamedTemporaryFile(suffix=".png") as f:
    urllib.request.urlretrieve(url, f.name)
    img = mpimg.imread(f.name)
    ax1.imshow(img, extent=[xmin, xmax, ymin, ymax])
    ax1.set_title("Landsat 7 Image")
    ax1.set_xlim(xmin, xmax)
    ax1.set_ylim(ymin, ymax)

    # Add red rectangle for region of interest
    width = rxmax - rxmin
    height = rymax - rymin
    rect = patches.Rectangle((rxmin, rymin), width, height,
                             linewidth=2, edgecolor='red', facecolor='none')
    ax1.add_patch(rect)
    ax1.set_xlabel("Longitude")
    ax1.set_ylabel("Latitude")

# Plot reflectance
ax2.plot(labels, reflectance, marker='o', color='blue')
ax2.set_title("Spectral Reflectance Curve")
ax2.set_ylabel("Reflectance")
ax2.set_xticks(range(len(labels)))
ax2.set_xticklabels(labels, rotation=45, ha='right')
ax2.grid(True)

plt.tight_layout()
plt.show()
```

```python
import tempfile
import urllib.request
import ee
import geemap
import matplotlib.image as mpimg
import matplotlib.patches as patches
import matplotlib.pyplot as plt

# Initialize Earth Engine
# ee.Initialize(project="earthengine-ml-testing")
ee.Initialize(project='earthenginetest-504004')


# 1. Define ROI: Greater Christchurch, New Zealand
christchurch_roi = ee.Geometry.Rectangle([172.45, -43.62, 172.75, -43.42])

# 2. Filter Sentinel-2 Level-2A (Surface Reflectance) Image Collection
s2_collection = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(christchurch_roi)
    .filterDate("2023-01-01", "2023-12-31")
    .sort("CLOUDY_PIXEL_PERCENTAGE")
)

# Select best scene (least cloud cover)
image = s2_collection.first()

# Quality Assessment: Check cloud coverage over scene and ROI
scene_cloud_pct = image.get("CLOUDY_PIXEL_PERCENTAGE").getInfo()

# Mask clouds using QA60 band for ROI precision calculation
qa = image.select("QA60")
cloud_bit_mask = 1 << 10
cirrus_bit_mask = 1 << 11
mask = (
    qa.bitwiseAnd(cloud_bit_mask)
    .eq(0)
    .And(qa.bitwiseAnd(cirrus_bit_mask).eq(0))
)
masked_img = image.updateMask(mask)

total_pixels = (
    ee.Image.pixelArea()
    .clip(christchurch_roi)
    .reduceRegion(
        reducer=ee.Reducer.count(),
        geometry=christchurch_roi,
        scale=10,
        maxPixels=1e9,
    )
    .getInfo()["area"]
)

valid_pixels = (
    masked_img.select("B2")
    .reduceRegion(
        reducer=ee.Reducer.count(),
        geometry=christchurch_roi,
        scale=10,
        maxPixels=1e9,
    )
    .getInfo()["B2"]
)

roi_cloud_pct = 100 * (1.0 - (valid_pixels / total_pixels))

print(f"Metadata Scene Cloud Cover: {scene_cloud_pct:.2f}%")
print(f"ROI Specific Calculated Cloud Cover: {roi_cloud_pct:.2f}%")

# 3. Define Representative Sampling Coordinates for 5 Land Cover Classes
samples = {
    "Urban": ee.Geometry.Point([172.6362, -43.5321]).buffer(50),
    "Forest": ee.Geometry.Point([172.7011, -43.5852]).buffer(50),
    "Grassland": ee.Geometry.Point([172.5800, -43.5500]).buffer(50),
    "Water": ee.Geometry.Point([172.7200, -43.5200]).buffer(50),
    "Bare Soil": ee.Geometry.Point([172.5200, -43.6000]).buffer(50),
}

# Sentinel-2 Bands & Scale Factor (1/10000 for S2 SR)
s2_bands = ["B2", "B3", "B4", "B8", "B11", "B12"]
band_labels = [
    "Blue (B2)",
    "Green (B3)",
    "Red (B4)",
    "NIR (B8)",
    "SWIR1 (B11)",
    "SWIR2 (B12)",
]
colors = {
    "Urban": "#d62728",
    "Forest": "#2ca02c",
    "Grassland": "#bcbd22",
    "Water": "#1f77b4",
    "Bare Soil": "#8c564b",
}

# Extract mean spectral curves
spectral_curves = {}
for name, geom in samples.items():
    stats = image.select(s2_bands).reduceRegion(
        reducer=ee.Reducer.mean(), geometry=geom, scale=10, maxPixels=1e6
    )
    # Scale raw values to surface reflectance (0-1)
    vals = [stats.getInfo()[b] * 0.0001 for b in s2_bands]
    spectral_curves[name] = vals

# 4. Generate Publication-Quality Figure
plt.style.use("seaborn-v0_8-whitegrid" if "seaborn-v0_8-whitegrid" in plt.style.available else "default")
fig = plt.figure(figsize=(12, 10), dpi=300)

# Panel A: Map Preview with ROI and Sample Location Boxes
ax1 = fig.add_subplot(2, 1, 1)

# Fetch Thumbnail image over Christchurch
coords = christchurch_roi.coordinates().getInfo()[0]
lons = [p[0] for p in coords]
lats = [p[1] for p in coords]
xmin, xmax, ymin, ymax = min(lons), max(lons), min(lats), max(lats)

thumb_params = {
    "dimensions": 800,
    "region": christchurch_roi,
    "format": "png",
    "bands": ["B4", "B3", "B2"],
    "min": 0,
    "max": 3000,
}
url = image.getThumbURL(thumb_params)

with tempfile.NamedTemporaryFile(suffix=".png") as f:
    urllib.request.urlretrieve(url, f.name)
    img = mpimg.imread(f.name)
    ax1.imshow(img, extent=[xmin, xmax, ymin, ymax])

# Overlay Sample Area Markers/Boxes
for name, geom in samples.items():
    g_coords = geom.bounds().coordinates().getInfo()[0]
    g_lons = [p[0] for p in g_coords]
    g_lats = [p[1] for p in g_coords]
    rect_w = max(g_lons) - min(g_lons)
    rect_h = max(g_lats) - min(g_lats)
    rect = patches.Rectangle(
        (min(g_lons), min(g_lats)),
        rect_w,
        rect_h,
        linewidth=2,
        edgecolor=colors[name],
        facecolor="none",
        label=f"Sample: {name}",
    )
    ax1.add_patch(rect)
    ax1.text(
        min(g_lons),
        max(g_lats) + 0.005,
        name,
        color=colors[name],
        fontsize=10,
        weight="bold",
    )

ax1.set_title(
    "A) Sentinel-2 True Colour Composite & Land Cover Sample Locations (Christchurch, NZ)",
    fontsize=13,
    weight="bold",
)
ax1.set_xlabel("Longitude (°E)", fontsize=11)
ax1.set_ylabel("Latitude (°S)", fontsize=11)

# Panel B: Stacked Spectral Reflectance Curves
ax2 = fig.add_subplot(2, 1, 2)
for name, vals in spectral_curves.items():
    ax2.plot(
        band_labels,
        vals,
        marker="o",
        linewidth=2.5,
        markersize=7,
        color=colors[name],
        label=name,
    )

ax2.set_title(
    "B) Comparative Sentinel-2 Spectral Reflectance Curves across Land Cover Types",
    fontsize=13,
    weight="bold",
)
ax2.set_xlabel("Sentinel-2 Spectral Bands", fontsize=11)
ax2.set_ylabel("Surface Reflectance (0 - 1.0)", fontsize=11)
ax2.set_ylim(0, 0.6)
ax2.grid(True, linestyle="--", alpha=0.7)
ax2.legend(title="Land Cover Class", fontsize=10, title_fontsize=11, loc="upper left")

plt.tight_layout()
plt.savefig("Christchurch_Spectral_Reflectance_Analysis.png", dpi=300, bbox_inches="tight")
plt.show()
```
